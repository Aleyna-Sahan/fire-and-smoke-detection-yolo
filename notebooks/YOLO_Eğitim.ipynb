VERİNİ YOLARI BOZUK ONLARI DÜZELTTİK

```python
import os
import yaml
import glob

# 1. İndirdiğin veri setinin ana dizini
base_dir = "/kaggle/input/smoke-fire-detection-yolo"

# 2. 'train/images' klasörünün tam olarak nerede olduğunu otomatik bulalım
train_search = glob.glob(f"{base_dir}/**/train/images", recursive=True)

if train_search:
    train_dir = train_search[0]
    # Veri setinin gerçek kök dizini (train/images klasörünün iki üstü)
    dataset_root = os.path.dirname(os.path.dirname(train_dir))

    # 3. Orijinal data.yaml dosyasını oku
    with open(os.path.join(base_dir, "data.yaml"), 'r') as f:
        data = yaml.safe_load(f)

    # 4. Yolları Colab ortamına uygun şekilde (göreli yollarla) güncelle
    data['path'] = dataset_root
    data['train'] = 'train/images'
    data['val'] = 'val/images'
    if 'test' in data:
        data['test'] = 'test/images'

    # 5. Düzeltilmiş ayarları Colab ana dizinine yeni bir dosya olarak kaydet
    new_yaml_path = "/content/custom_data.yaml"
    with open(new_yaml_path, 'w') as f:
        yaml.dump(data, f)

    print(f"✅ Harika! Yollar düzeltildi ve '{new_yaml_path}' dosyası başarıyla oluşturuldu.")
else:
    print("❌ Klasör yapısı bulunamadı. Veri seti eksik veya hatalı inmiş olabilir.")
```
