```python
%pip install -q --upgrade ultralytics
from ultralytics import YOLO
import gc
import torch

# 1. Modeli yükle
model = YOLO("/content/drive/MyDrive/YOLO_Ates_Duman/train-2/weights/best.pt")

# 2. Test videosu (YouTube linki)
test_video = "https://www.youtube.com/watch?v=Vu_Q0sflTeY"

# 3. RAM temizliği (Başlamadan önce evi temizleyelim)
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("🔥 Video işleniyor ve doğrudan Google Drive'a aşama aşama kaydediliyor...")

# 4. Canlı Testi Başlat! (RAM optimizasyonlu ve Drive Hedefli)
results = model.predict(
    source=test_video,
    conf=0.5,
    save=True,
    stream=True,        # 🔥 Kare kare işle, RAM'de biriktirme
    imgsz=320,          # Daha düşük çözünürlük
    vid_stride=2,       # Her 2 kareden 1'ini işle
    half=True,          # FP16 hassasiyet
    verbose=False,      # Gereksiz logları kapat
    project="/content/drive/MyDrive/TOTTİK/AleynaMakina/YOLO_Ates_Duman",
    name="Test_Ciktilari",
    exist_ok=True
)

# 5. Sonuçları işle ve belleği temiz tut
for i, r in enumerate(results):
    if i % 100 == 0:  # Her 100 karede bir bilgi ver
        print(f"İşleniyor... Kare: {i}")

    # Döngü içindeki geçici belleği temizle
    del r
    if i % 50 == 0:
        gc.collect()

print("✅ Test bitti! İşlenmiş video doğrudan Drive'daki 'Test_Ciktilari' klasörüne kaydedildi.")
```
